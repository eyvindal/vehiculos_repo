```python
import pandas as pd
import plotly.graph_objects as go
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#se carga el conjunto de datos como un dataframe de pandas
df = pd.read_csv(r"C:\python\sprint7_tf\smartphones_repo\data\vehicles_us.csv")
```

```python
#visualizacion de los datos
print(df.head())
```

```python
#exploracion de los datos
df.info()
df.describe()
```

```python
#Dado que se trabaja con un set de datos que suponemos real prefiero no imputar los valores ausentes
#Busqueda de duplicados explicitos
print(df.duplicated().sum())
```

```python
#Inicia la visualizacion de datos

#creacion de histogramas
fig = go.Figure(data=[go.Histogram(x=df['model_year'])])
fig.update_layout(title_text='Distribución de los ejemplares por año')
fig.show()
```

```python
sns.histplot(data=df,x=df["type"])
plt.title("Distribución por tipo de vehículo")
plt.xlabel("Tipo de vehículo")
plt.ylabel("Cantidad de vehículos")
plt.xticks(rotation=50)
plt.show()
```

```python
#cracion de boxplot
sns.boxplot(data=df,x=df["condition"], y=df["price"])
plt.title("Relación entre la condición del vehículo y su precio") 
plt.xlabel("Condición del vehículo")
plt.ylabel("Precio (USD)")
plt.show()
```

```python
sns.lineplot(data=df,x=df["model_year"], y=df["price"])
plt.title("Relación entre el año del modelo y el precio del vehículo") 
plt.xlabel("Año del modelo")
plt.ylabel("Precio (USD)")
plt.xticks(rotation=90)
plt.show()
```

```python
#creacion de graficos de dispersion
sns.scatterplot(data=df,x=df["odometer"], y=df["price"])
plt.title("Relación entre el odómetro y el precio del vehículo")
plt.xlabel("Odómetro en mill. de millas")
plt.ylabel("Precio del vehículo")
plt.show()
```

```python
sns.scatterplot(data=df,x=df["model_year"], y=df["price"])
plt.title("Relación entre el año del modelo y el precio del vehículo")
plt.xlabel("Año del modelo")
plt.ylabel("Precio del vehículo")
plt.show()
```

```python
#creación de grafico de barras
sns.countplot(data=df, x="fuel", hue="transmission")
plt.title("Vehículos por tipo de combustible y transmisión")
plt.xlabel("Tipo de conbustible")
plt.ylabel("Conteo de vehículos")
plt.show()
```
